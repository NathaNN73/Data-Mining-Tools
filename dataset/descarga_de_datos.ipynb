```python
import os
import shutil
```

# Carga de datos de la versión muestral de "Random Sample of NIH Chest X-ray Dataset"

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("nih-chest-xrays/sample")

print("Path to dataset files:", path)
```

```python
destino = "D:\Ciclo VIII\Data-Mining-Tools\TB\Datasets"
if os.path.exists(path):
    try:
        os.makedirs(os.path.dirname(destino), exist_ok=True)
        shutil.move(path, destino)
        print(f"Archivo movido a: {destino}")
    except Exception as e:
        print(f"Error al mover el archivo: {e}")
else:
    print("El archivo original no existe.")
```
