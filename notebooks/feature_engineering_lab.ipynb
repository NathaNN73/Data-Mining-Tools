```python
precio_medianas = data.groupby("StockCode")["UnitPrice"].median().to_dict()

def marcar_descuento_alto(row):
    codigo = row["StockCode"]
    precio_actual = row["UnitPrice"]
    precio_normal = precio_medianas.get(codigo)
    
    if precio_normal is None or precio_normal == 0:
        return 0
    
    if precio_actual < 0.5 * precio_normal:
        return 1
    else:
        return 0

data["Descuento_alto"] = data.apply(marcar_descuento_alto, axis=1)
```

## Tarea 3

```python
class Preprocessor(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.precio_medianas_ = {}

    def fit(self, X, y=None):
        self.precio_medianas_ = X.groupby("StockCode")["UnitPrice"].median().to_dict()
        return self

    def transform(self, X):
        X = X.copy()

        X['Valor_total'] = X['Quantity'] * X['UnitPrice']
        X['Mes'] = X['InvoiceDate'].dt.month
        X['Dia_semana'] = X['InvoiceDate'].dt.dayofweek

        def marcar_descuento_alto(row):
            codigo = row["StockCode"]
            precio_actual = row["UnitPrice"]
            precio_normal = self.precio_medianas_.get(codigo)

            if precio_normal is None or precio_normal == 0:
                return 0
            if precio_actual < 0.5 * precio_normal:
                return 1
            else:
                return 0

        X["Descuento_alto"] = X.apply(marcar_descuento_alto, axis=1)

        return X
```
